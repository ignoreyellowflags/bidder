```python
'''
Чтобы обновить состояние/добавить новую актуальную статистику - используйте метод `update`
Инициализация объявления
'''
AdExp = Ad(
    accountId='my-account',
    adId='my-ad',
    impression=0,
    click=0,
    expenditure=0,
    dateStart="2026-03-02 00:00:00",
    dateEnd="2026-03-08 23:00:00",
    touchDate="2026-03-02 00:00:00"
)
```

```python
# Проверка статистик
print(f"impression: {AdExp.impression}, click: {AdExp.click}, expenditure: {AdExp.expenditure}")
```

```python
# Прошел час и `touchDate` изменился на 1 час
AdExp.update(impression=3000, click=25, expenditure=3, touchDate="2026-03-02 01:00:00")
```

```python
'''
Когда класс инициализируется на сусловиях холодного старта (`touchDate=None`), любая статистика отличная от нуля приведет к ошибке,
так как невозможно иметь просмотры/клики и траты не успев запуститься
'''
AdExp = Ad(
    accountId='my-account',
    adId='my-ad',
    impression=10,
    click=0,
    expenditure=0,
    dateStart="2026-03-02 00:00:00",
    dateEnd="2026-03-08 23:00:00",
    touchDate=None
)
```

```python
'''
При попытке обновить данные ретроспективно (`touchDate` метода  update "меньше" `touchDate` класса из __init__) 
- так же приведет к ошибке
'''
AdExp = Ad(
    accountId='my-account',
    adId='my-ad',
    impression=1000,
    click=20,
    expenditure=4,
    dateStart="2026-03-02 00:00:00",
    dateEnd="2026-03-08 23:00:00",
    touchDate="2026-03-02 04:00:00"
)
```

```python
AdExp.update(impression=3000, click=25, expenditure=3, touchDate="2026-03-02 02:00:00")
```

```python
# Bidder
'''
`Bidder` определяется двумя сущностями:
    - классом `Ad` (рекламное объявление), чьей ставкой предстоит управлять
    - корридором CPM
'''
# Шаг 1. Определяем рекламное объявление
AdExp = Ad(
    accountId='my-account',
    adId='my-ad',
    impression=1000,
    click=20,
    expenditure=4,
    dateStart="2026-03-02 00:00:00",
    dateEnd="2026-03-08 23:00:00",
    touchDate="2026-03-02 04:00:00"
)
# Шаг 2. Инициализируем класс Bidder
bidder = Bidder(ad=AdExp, cpm=(1, 4))
```

```python
bidder.pace(weekEndTimeOffActivity=False, nightTimeOffActivity='early', freq='1hour')
```
